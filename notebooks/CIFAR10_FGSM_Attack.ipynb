# FGSM Attack on CIFAR-10

This notebook demonstrates the **FGSM** attack on the CIFAR-10 dataset using `sleight`.

```python
import numpy as np
import matplotlib.pyplot as plt

from sleight.data import get_dataset
from sleight.models import get_cifar10_cnn_model
from sleight.attacks import fgsm_attack
from sleight.evaluation import evaluate_robustness, epsilon_sweep, plot_accuracy_vs_epsilon
```

```python
# Load CIFAR-10
(x_train, y_train), (x_test, y_test) = get_dataset("cifar10", one_hot=True)
y_test_int = np.argmax(y_test, axis=1)
print(f"Train: {x_train.shape}, Test: {x_test.shape}")
```

```python
# Build and train model
model = get_cifar10_cnn_model()
model.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.1, verbose=1)
```

```python
# Epsilon sweep
x_sub = x_test[:300]
y_sub_int = y_test_int[:300]
epsilons = [0.0, 0.01, 0.03, 0.05, 0.1, 0.15]

results = epsilon_sweep(model, x_sub, y_sub_int, fgsm_attack, epsilons)
for r in results:
    print(f"eps={r['epsilon']:.3f}  clean={r['clean_accuracy']:.3f}  adv={r['adversarial_accuracy']:.3f}")
```

```python
# Plot
plot_accuracy_vs_epsilon(results, title="FGSM on CIFAR-10", show=True)
```
